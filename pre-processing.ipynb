```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("data/data_gempa_processed.csv")
df.head()
```

```python
df = df.reset_index(drop=True)
df["id"] = df.index + 1

df = df.drop(columns=["province_id"])
```

```python
df = df.rename(columns={"place": "province"})
```

```python
print('Jumlah province_id unik:', df['id'].nunique())
print('\nContoh 10 province_id unik:')
print(df['id'].unique()[:100])
```

```python
df.head()
```

```python
df = df[
    (df["magnitude"] > 0) &
    (df["depth"] >= 0) &
    (df["latitude"].between(-90, 90)) &
    (df["longitude"].between(-180, 180))
]
```

Normalisasi Waktu

```python
df["time"] = pd.to_datetime(df["time"], errors="coerce")
df["date"] = df["time"].dt.date
df["year"] = df["time"].dt.year
df["month"] = df["time"].dt.month
df["month_name"] = df["time"].dt.month_name()
df.head()
```

kategori Depth

```python
def depth_category(depth):
    if pd.isna(depth):
        return None
    elif depth < 60:
        return "dangkal"
    elif depth <= 300:
        return "menengah"
    else:
        return "dalam"
    
df["depth_category"] = df["depth"].apply(depth_category)
```

```python
df.head()
```

standasisasi nama provinsi

```python
df["place"] = (
    df["place"]
    .str.strip()
    .str.title()
)
ordered_cols = [
    "id",
    "province_id",
    "place",
    "time",
    "date",
    "month",
    "month_name",
    "year",
    "latitude",
    "longitude",
    "magnitude",
    "depth",
    "depth_category",
    "fault_megathrust_id"
]

df = df[ordered_cols]
df.head()
```

```python
df.to_csv("data/data_gempa_preprocessed.csv", index=False)
```
