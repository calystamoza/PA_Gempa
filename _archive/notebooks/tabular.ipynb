```python
import pandas as pd
from llama_index.core.query_engine import PandasQueryEngine
from llama_index.llms.groq import Groq

# 1. Load data dari path kamu
path_data = "data/data_gempa_processed.csv"
df = pd.read_csv(path_data)

# 2. Inisialisasi LLM (Gunakan model yang sama dengan TA kamu)
# Masukkan API Key Groq kamu di sini
llm = Groq(model="llama-3.3-70b-versatile", api_key="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")

# 3. Buat Query Engine
# verbose=True supaya kita bisa lihat "proses berpikir" AI saat membuat kode Python
query_engine = PandasQueryEngine(df=df, llm=llm, verbose=True)

# 4. Uji Coba Pertanyaan Statistik
response = query_engine.query(
    "Berapa rata-rata magnitudo gempa di Jawa Barat selama tahun 2024?"
)

print(f"Jawaban AI: {response}")
```
